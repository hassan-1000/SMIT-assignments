```python
guests = []
for name in range(10):
    name = input("Who are you inviting to your party")

    if name in guests:
        print("This person is already added in the list")

    else:
        guests.append(name)
        print("Guest added successfully")

print("Final guest list")
print(guests)
```

```python
Discarded_dishes = "fish"
menu = []
for dish in range(5):
    dish = input("What dish would you add in the menu")

    if Discarded_dishes  in dish :
        print("This dish is prohibited in the menu")

    else:
        menu.append(dish)
        print("Dish added successfully")

print("Final menu list")
print(menu)
```

```python
menu
```

```python
menu_prices = {}
for dish in menu:
    price = int(input(f"Input Price for {dish}:"))

    menu_prices[dish] = price
```
