```python
books = []

def add_book():
    bookID = input("Enter BookID: ")
    title = input("Enter title: ")
    author_name = input("Enter author's name: ")
    number_copies = int(input("Enter amount of copies: "))

    new_books = {
        "BookID": bookID,
        "Title": title,
        "Author_name": author_name,
        "Number_of_copies": number_copies
    }

    books.append(new_books)
    print("Book added successfully!")


def display_record():
    if not books:
        print("Records are empty")
    else:
        for new_books in books:
            print("-------------------------------")
            print("Book:")
            for key, value in new_books.items():
                print(f"{key}: {value}")
            print("-------------------------------")


def search_books(BookID):
    if not books:
        print("Unable to search in empty records")
    else:
        found = False
        for new_books in books:
            if new_books["BookID"] == BookID:
                for key, value in new_books.items():
                    print(f"{key}: {value}")
                found = True
                break

        if not found:
            print("Book not found")


def update_Number_of_copies(BookID):
    if not books:
        print("Unable to update copies in empty records")
    else:
        for index, new_books in enumerate(books):
            if new_books["BookID"] == BookID:
                new_copies = int(input("Enter updated number of copies: "))
                books[index]["Number_of_copies"] = new_copies
                print("Copies updated successfully")
                return

        print("Book not found")


def delete_book(BookID):
    if not books:
        print("Unable to delete from empty records")
    else:
        for new_books in books:
            if new_books["BookID"] == BookID:
                books.remove(new_books)
                print("Record successfully deleted")
                return

        print("Book not found")

def issue_books(BookID):
    if not books:
        print("Unable to search in empty records")
    else:
        found = False
        for new_books in books:
            if new_books["BookID"] == BookID:
                new_books["Number_of_copies"] -= 1 
                found = True
                break


def return_books(BookID):
    if not books:
        print("Unable to search in empty records")
    else:
        found = False
        for new_books in books:
            if new_books["BookID"] == BookID:
                new_books["Number_of_copies"] += 1 
                found = True
                break
        if not found:
            print("Book not found")


while True:
    print("\n----- Library Management System -----")
    print("1. Press c to add record")
    print("2. Press r to retrieve records")
    print("3. Press s to search record")
    print("4. Press u to update copies")
    print("5. Press d to delete record")
    print("6. Press i to issue book")
    print("7. Press x to return")
    print("6. Press e to exit")

    option = input("Enter your choice: ").lower()

    if option == "c":
        add_book()
    elif option == "r":
        display_record()
    elif option == "s":
        search_id = input("Enter BookID to search: ")
        search_books(search_id)
    elif option == "u":
        search_id = input("Enter BookID to update copies: ")
        update_Number_of_copies(search_id)
    elif option == "d":
        search_id = input("Enter BookID to delete: ")
        delete_book(search_id)
    elif option == "e":
        print("Exiting Library Management System")
        break
    elif option == "i":
        search_id = input("Enter BookID to issue book: ")
        issue_books(search_id)

    elif option == "x":
        search_id = input("Enter BookID to return book: ")
        return_books(search_id)
        
    else:
        print("Invalid choice")
```
